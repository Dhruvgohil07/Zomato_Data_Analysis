# Feature Engineering on the cleaned Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Output/df_india_cleaned', encoding='latin-1')
df.head()
```

### Per-restaurant cuisines:
This are the number of cuisines that the restaurants servers.
### Primary cuisines:
This are the restaurant's main cuisine, zomato lists the dominant cuisines first.

```python
df.columns
```

```python
df['Cuisines']
```

```python
cuisines_split=df['Cuisines'].str.split(',')
df['Cuisines Count']=cuisines_split.apply(len)
df['Primary Cuisine']=cuisines_split.apply(lambda x: x[0])
```

```python
df.columns
```

```python
df[['Restaurant Name', 'Cuisines Count', 'Primary Cuisine']]
```

```python
Average_muli_cuisines=df[df['Cuisines Count']>1]['Average Cost for two'].mean()
print(f"Average cost for Multi cuisine restaurants: {Average_muli_cuisines}")
Average_single_cuisine=df[df['Cuisines Count']==1]['Average Cost for two'].mean()
print(f"Average cost for two peoples in the restaurants that are serving single primary cuisine: {Average_single_cuisine}")
```

### Observation:
From the above result we can say that the average cost for two people are greater in Multi-Cuisine restaurants as compared to the cost for two people in a restaurant that serves only one cuisine.
