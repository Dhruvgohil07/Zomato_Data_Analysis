# Feature Engineering on the cleaned Data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
df=pd.read_csv('Output/df_india_cleaned', encoding='latin-1')
df.head()
```

### Per-restaurant cuisines:
This are the number of cuisines that the restaurants servers.
### Primary cuisines:
This are the restaurant's main cuisine, zomato lists the dominant cuisines first.

```python
df.columns
```

```python
df['Cuisines']
```

```python
cuisines_split=df['Cuisines'].str.split(',')
df['Cuisines Count']=cuisines_split.apply(len)
df['Primary Cuisine']=cuisines_split.apply(lambda x: x[0])
```

```python
df.columns
```

```python
df[['Restaurant Name', 'Cuisines Count', 'Primary Cuisine']]
```

```python
Average_muli_cuisines=df[df['Cuisines Count']>1]['Average Cost for two'].mean()
print(f"Average cost for Multi cuisine restaurants: {Average_muli_cuisines}")
Average_single_cuisine=df[df['Cuisines Count']==1]['Average Cost for two'].mean()
print(f"Average cost for two peoples in the restaurants that are serving single primary cuisine: {Average_single_cuisine}")
```

### Observation:
From the above result we can say that the average cost for two people are greater in Multi-Cuisine restaurants as compared to the cost for two people in a restaurant that serves only one cuisine.

```python
#Now we are converting the feature like has table booking and has Online Delivery to Binary flag like 0 and 1
df[['Restaurant Name', 'Has Table booking', 'Has Online delivery']]
```

```python
df['Has Online delivery']=df['Has Online delivery'].map({
    'Yes': 1,
    'No':0
})
df['Has Table booking']=df['Has Table booking'].map({
    'Yes':1,
    'No':0
})
```

```python
df[['Restaurant Name', 'Has Table booking', 'Has Online delivery']]
```

## Skewness:
It tell where the "extreme" values live:

- Positive skewnesss-Extreme values are on the right side

- Negetive skewness-Extreme values are on the left side

- Zero-This means data is balanced around the center

```python
#checking the skewness of some of the columns:
votes_skewness=df['Votes'].skew()
print(votes_skewness)
```

### Observation:
We can see that the Votes are highly skew on the right side. Vast majority of the restaurants have very few votes. A tiny number of viral/popular restaurants have thousands of votes, creating massive right tail.


```python
#Since the votes are heavily skewed on the right side applying the transformation on the votes column, this will make the charts, correlation and later models more reliable
df['Log_Votes']=np.log1p(df['Votes'])
```
